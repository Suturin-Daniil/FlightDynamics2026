```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.interpolate import interp1d
```

```python
fig, ax = plt.subplots(1,1, figsize=(7,4))
b = np.linspace(-10, 10, 22)
Cy = -0.01*b
Cy[-1] = Cy[-1]*0.65
Cy[-2] = Cy[-2]*0.8 
Cy[-3] = Cy[-3]*0.95

Cy[0] = Cy[0]*0.65
Cy[1] = Cy[1]*0.8 
Cy[2] = Cy[2]*0.95

f = interp1d(b, Cy, kind='cubic')
ax.scatter(b, Cy, s=3, c='black')
ax.plot(b, f(b), linewidth = 1)
ax.spines['bottom'].set_position(('data', 0))
ax.set_yticks([])
ax.set_ylabel(r'$C_{Y}$')
ax.set_xlabel(r'$\beta$')
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
ax.annotate('Срыв потока \n на руле направления', (5, -0.08), (2, 0.06))
ax.annotate(
    "", 
    xy=(-8, 0.075),          # Arrow head destination
    xytext=(4, 0.05),      # Arrow tail origin
    arrowprops=dict(arrowstyle="->", color="red", lw=1)
)
ax.annotate(
    "", 
    xy=(8, -0.075),          # Arrow head destination
    xytext=(4, 0.05),      # Arrow tail origin
    arrowprops=dict(arrowstyle="->", color="red", lw=1)
)

ax.vlines(-7, 0.02, 0.07, color='black', linewidth=1)
ax.hlines(0.02, -7, -2, color='black', linewidth=1)
ax.text(-10, 0.04, r'$\Delta C_{Y} < 0$')
ax.text(-7, 0.01, r'$\Delta \beta > 0$')
ax.text(-6.8, 0.03, r'$\frac{\partial C_{Y}}{\partial \beta} < 0$', size=14)

Cy1 = Cy + 0.02 # приращение засчет оклоненя руля направления
Cy2 = Cy - 0.02 # приращение засчет оклоненя руля направления
ax.plot(b,Cy1, linewidth = 1, color='green')
ax.annotate(
    "", 
    xy=(5, -0.028),          # Arrow head destination
    xytext=(5, -0.052),      # Arrow tail origin
    arrowprops=dict(arrowstyle="<->", color="black", lw=1)
)
ax.text(2.5, -0.06, r"$\delta_{r} = +5\degree$")

plt.savefig('CY(beta).png', dpi=300)
```
